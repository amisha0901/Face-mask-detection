```python
from tensorflow.keras.preprocessing import image
```

```python
image_size = (64,64)
class_indices = ['Bear', 'Bird', 'Cat', 'Cow', 'Deer', 'Dog', 'Dolphin', 'Elephant', 'Giraffe', 'Horse', 'Kangaroo', 'Lion', 'Panda', 'Tiger', 'Zebra']
```

```python
def get_prediction(image_path, model):
    input_image = image.load_img(image_path, target_size = image_size)
    image_array = image.img_to_array(input_image)/255  
    test_array = image_array.reshape(1, 64, 64, 3)
    prediction = model.predict(test_array)
    output = sorted(list(zip(list(class_indices), list(prediction[0]))), key = lambda x:x[1])[-1]
    
    return output
```
